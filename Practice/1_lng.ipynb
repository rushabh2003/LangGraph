```python
from langgraph.graph import StateGraph , START , END
from typing import TypedDict
from langchain_groq import ChatGroq
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
class sqsum(TypedDict):
    val1 : int
    val2 : int
    res : int
```

```python
def calsum(state : sqsum)->sqsum:
    v1 = state['val1']
    v2 = state['val2']
    result = v1 + v2

    state['res'] = result

    return state

```

```python
graph = StateGraph(sqsum)

graph.add_node('calsum',calsum)

graph.add_edge(START,'calsum')
graph.add_edge('calsum',END)

workflow = graph.compile()
```

```python
workflow.invoke({'val1':2 , 'val2':3})
```
